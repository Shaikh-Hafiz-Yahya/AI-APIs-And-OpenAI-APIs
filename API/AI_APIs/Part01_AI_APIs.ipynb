# API Keys aur Environment Variables

## API Keys Kya Hain?

API keys special passwords ki tarah hoti hain jo aap ko kisi API service ko access aur use karne ki permission deti hain. Jab aap kisi AI API ko request bhejte hain, to aap ko apni API key include karni hoti hai taake service ye jaan sake:

1. Aap kaun hain

2. Kya aap ko service use karne ki permission hai

3. Aap ke usage ko kaise track karna hai (bohat si APIs par usage limits ya costs hoti hain)

**Important:** API keys ko secret rakhna chahiye! Agar kisi ko aap ki API key mil jaye, to woh aap ban kar service use kar sakta hai, jis ki wajah se aap ko potentially paise dene par sakte hain ya aap ka quota khatam ho sakta hai.

## API Keys ke liye `.env` Files ka Istemaal

`.env` file ek simple text file hoti hai jo environment variables (jaise API keys) ko store karti hai, jinhein aap ka program access kar sakta hai. `.env` file ka istemaal kai reasons ki wajah se ek best practice hai:

1. **Security**: Sensitive information ko aap ke code se bahar rakhti hai

2. **Portability**: Different environments mein different keys use karna easy banati hai

3. **Version Control**: API keys ko GitHub ya kisi doosri repository par upload hone se prevent karti hai

Chaliye dekhte hain ke `.env` file kaise create aur use karte hain:

### Step 1: `.env` File Create Karein

Ek new file banayein jiska naam `.env` ho (shuru mein dot ko note karein) aur isay apni notebook ya Python script ke same folder mein rakhein. Apni API keys is tarah add karein:

OPENAI_API_KEY=your_openai_api_key_here

GOOGLE_API_KEY=your_google_api_key_here


`your_openai_api_key_here` aur `your_google_api_key_here` ko apni actual API keys se replace karein.

**Important:** Is file ko kabhi bhi doosron ke sath share na karein aur na hi ise public repositories par upload karein!

### Step 2: `python-dotenv` Install Karein

Humein apni `.env` file ko read karne ke liye `python-dotenv` naam ke ek package ki zaroorat hai. Chaliye ise install karte hain:

```python
# Install the python-dotenv package
#!pip install python-dotenv
```

### Step 3: Environment Variables Load Karein

Ab hum apni API keys ko `.env` file se load kar sakte hain:

```python
import os
from dotenv import load_dotenv

# .env file se environment variables load karein
load_dotenv()

# Naam ke zariye API key hasil karne ke liye function
def get_api_key(key_name):

    try:
        key = os.getenv(key_name)

        if key:
            print(f"{key_name} successfully load ho gayi.")
        else:
            print(f"{key_name} nahi mili. Apni .env file check karein.")
        return key
    
    except Exception as error:
        print(f"{key_name} load karte waqt kuch ghalat ho gaya: {error}")
        return None

# API keys load karein
print('_'*100)
openai_api_key = get_api_key("OPENAI_API_KEY")
google_api_key = get_api_key("GOOGLE_API_KEY")
print('_'*100)
```

### Upar Wale Code Mein Kya Ho Raha Hai?

1. Hum `os` module import karte hain, jo humein environment variables access karne ki facility deta hai

2. Hum `dotenv` module se `load_dotenv` function import karte hain

3. Hum `load_dotenv()` call karte hain taake hamari `.env` file mein mojood tamam variables environment mein load ho jayein

4. Hum `os.getenv("VARIABLE_NAME")` use karte hain taake har API key ko hasil kar saken

5. Hum check karte hain ke API keys successfully load hui hain ya nahi, aur confirmation message print karte hain

### Alternative: Environment Variables Directly Use Karna

Agar aap `.env` file use nahi kar sakte (for example, kuch online environments mein), to aap environment variables ko directly apne code mein set kar sakte hain, lekin ye method **kam secure** hai:

```python
# Sirf is method ko testing ke liye use karein ya jab `.env` files ka option available na ho

# `'your_api_key_here'` ko apni actual API keys se replace karein

# import os
os.environ["OPENAI_API_KEY"] = "your_api_key_here"  # Production code ke liye recommended nahi hai
os.environ["GOOGLE_API_KEY"] = "your_api_key_here"  # Production code ke liye recommended nahi hai
```

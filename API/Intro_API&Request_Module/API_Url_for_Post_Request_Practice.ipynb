# **`API URL FOR POST REQUEST`**

```python
import requests
```

```python
Api_Url = "https://jsonplaceholder.typicode.com/posts"

# Data to send request
PostData ={

    'Title':'New Post',
    'Body':'This is a new post created using the request module',
    'Id':1

}

ApiResponse = requests.post(Api_Url , json=PostData)

print(f"Status Code: {ApiResponse.status_code} = Response Successfull!")
print(ApiResponse.json())
```

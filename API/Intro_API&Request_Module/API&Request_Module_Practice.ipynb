```python
import requests
```

```python
apiUrl = "https://en.wikipedia.org/wiki/Natural_language_processing"

responseApi = requests.get(apiUrl)

print(f"{responseApi.headers}\n")
print(f"Content Length={responseApi.headers.get("content-length")}")
print(f"Status Code={responseApi.status_code}")
print(f"Header Length={len(responseApi.headers)}")
```
